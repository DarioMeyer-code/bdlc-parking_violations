```python
df_2023 = spark.read.csv(
    "hdfs:///parking_violations/raw/2023/parking_violations_2023.csv",
    header=True,
    inferSchema=True
)

df_2024 = spark.read.csv(
    "hdfs:///parking_violations/raw/2024/parking_violations_2024.csv",
    header=True,
    inferSchema=True
)
```

```python
print("2023:", df_2023.count(), "Zeilen")
print("2024:", df_2024.count(), "Zeilen")
print("2025:", df_2025.count(), "Zeilen")
```

```python
print("2023 Spalten:", len(df_2023.columns))
print("2024 Spalten:", len(df_2024.columns))
print("2025 Spalten:", len(df_2025.columns))

print("2023 == 2024:", df_2023.columns == df_2024.columns)
print("2024 == 2025:", df_2024.columns == df_2025.columns)
```

```python
from pyspark.sql.functions import lit

df_2023_y = df_2023.withColumn("Fiscal Year", lit(2023))
df_2024_y = df_2024.withColumn("Fiscal Year", lit(2024))
df_2025_y = df_2025.withColumn("Fiscal Year", lit(2025))

df_all = df_2023_y.unionByName(df_2024_y).unionByName(df_2025_y)
```

```python
total_rows = df_all.count()

print("Gesamtzahl Zeilen:", total_rows)
```

```python
df_all.groupBy("Fiscal Year").count().orderBy("Fiscal Year").show()
```

```python
analysis_df = df_all.select(
    "Fiscal Year",
    "Summons Number",
    "Issue Date",
    "Violation Time",
    "Violation County",
    "Violation Precinct",
    "Street Name",
    "Vehicle Make",
    "Violation Code",
    "Violation Description"
)

analysis_df.show(5, truncate=False)
```

```python
from pyspark.sql.functions import col, sum as spark_sum

null_check = analysis_df.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in analysis_df.columns
])

null_check.show(truncate=False)
```

```python
clean_df = analysis_df.filter(
    col("Issue Date").isNotNull() &
    col("Violation Time").isNotNull() &
    col("Violation Code").isNotNull()
)

print("Clean rows:", clean_df.count())
print("Original rows:", analysis_df.count())
```

```python
original_rows = analysis_df.count()
clean_rows = clean_df.count()

removed_rows = original_rows - clean_rows
removed_percent = removed_rows / original_rows * 100

print("Entfernte Zeilen:", removed_rows)
print("Anteil entfernt (%):", removed_percent)
```

```python
from pyspark.sql.functions import to_date

clean_df = clean_df.withColumn(
    "issue_date_parsed",
    to_date(col("Issue Date"), "MM/dd/yyyy")
)

clean_df.select("Issue Date", "issue_date_parsed").show(10)
```

```python
clean_df.filter(col("issue_date_parsed").isNull()).count()
```

```python
from pyspark.sql.functions import regexp_extract

clean_df = clean_df.withColumn(
    "time_hour_raw",
    regexp_extract(col("Violation Time"), r"^(\d{2})\d{2}[AP]$", 1).cast("int")
).withColumn(
    "time_minute",
    regexp_extract(col("Violation Time"), r"^\d{2}(\d{2})[AP]$", 1).cast("int")
).withColumn(
    "time_ampm",
    regexp_extract(col("Violation Time"), r"^\d{4}([AP])$", 1)
)

clean_df.select(
    "Violation Time",
    "time_hour_raw",
    "time_minute",
    "time_ampm"
).show(10)
```

```python
from pyspark.sql.functions import when

clean_df = clean_df.withColumn(
    "violation_hour",
    when((col("time_ampm") == "A") & (col("time_hour_raw") == 12), 0)
    .when((col("time_ampm") == "A"), col("time_hour_raw"))
    .when((col("time_ampm") == "P") & (col("time_hour_raw") == 12), 12)
    .when((col("time_ampm") == "P"), col("time_hour_raw") + 12)
)

clean_df.select(
    "Violation Time",
    "time_hour_raw",
    "time_ampm",
    "violation_hour"
).show(20)
```
